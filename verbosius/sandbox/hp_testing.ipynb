```python
max_acc_ngram_range = dict()
for trial in study.trials:
    if max_acc_ngram_range.get(trial.params['ngram_range']) is None:
        max_acc_ngram_range[trial.params['ngram_range']] = trial.value
    elif max_acc_ngram_range[trial.params['ngram_range']] < trial.value:
        max_acc_ngram_range[trial.params['ngram_range']] = trial.value


max_acc_ngram_range
```

```python
max_acc_stopwords = dict()
for trial in study.trials:
    if max_acc_stopwords.get(trial.params['stop_words']) is None:
        max_acc_stopwords[trial.params['stop_words']] = [trial.value]
    else:
        max_acc_stopwords[trial.params['stop_words']].append(trial.value)

# display(max_acc_stopwords)
print('AVG with stopwords\t', np.mean(max_acc_stopwords['english']))
print('AVG without stopwords\t', np.mean(max_acc_stopwords[None]))
```
