```python
import subprocess

# Execute the command to get the nvidia-smi output
nvidia_smi_output = subprocess.check_output(["nvidia-smi", "-q", "-d", "TEMPERATURE"])

# Convert the output to a string
nvidia_smi_output_str = nvidia_smi_output.decode()

# Split the output into lines
lines = nvidia_smi_output_str.split('\n')

temps = []
for line in lines:
    if "GPU Current Temp" in line:
        temps.append(line)

print(temps)

# Get the temperature from the output
temperature = temps[0].split(':')[-1].split('C')[0].strip()

# get both temperatures
temperature1 = temps[0].split(':')[-1].split('C')[0].strip()
temperature2 = temps[1].split(':')[-1].split('C')[0].strip()

# Print the temperature
print(temperature1)
print(temperature2)


# make a function that does this

def get_gpu_temp():

    nvidia_smi_output = subprocess.check_output(["nvidia-smi", "-q", "-d", "TEMPERATURE"])

    nvidia_smi_output_str = nvidia_smi_output.decode()

    lines = nvidia_smi_output_str.split('\n')

    temps = []
    for line in lines:
        if "GPU Current Temp" in line:
            temps.append(line)

    temperature1 = temps[0].split(':')[-1].split('C')[0].strip()
    temperature2 = temps[1].split(':')[-1].split('C')[0].strip()

    return temperature1, temperature2

```
