```python
import matplotlib.pyplot as plt
import pickle

data = pickle.load(open('data.pkl', 'rb'))

train_acc = data["train"]
val_acc = data["val"]

plt.plot(train_acc)
plt.plot(val_acc)

plt.xlabel('Epoch')
plt.ylabel('Accuracy')
plt.legend(['Train', 'Validation'])
plt.title('Accuracy over 100 epochs')

# save plot
plt.savefig('plot.png')
```
