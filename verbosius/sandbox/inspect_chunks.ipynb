```python
import os
import gzip
import pickle

path = "/home/bigtech/data/verbosius/amazon/trainingdata/amazon_chunkdist_6165/"

trainpath = os.path.join(path, "train")
evalpath = os.path.join(path, "eval")
    
total = 0
for file in os.listdir(trainpath):
    with gzip.open(os.path.join(trainpath, file), "rb") as f:
        data = pickle.load(f)
        total += len(data)

total_1 = 0
for file in os.listdir(evalpath):
    with gzip.open(os.path.join(evalpath, file), "rb") as f:
        data = pickle.load(f)
        total_1 += len(data)

print("train:", total)
print("eval:", total_1)
print("total:", total + total_1)
```

```python
a = [[1,2,3], [4,5,6], [7,8,9]]
new_a = []
c_max_size = 5
c = []

for i, e in enumerate(a):

    rest = 5 - (len(c) + len(e))

    if rest >= 0 and len(c) < c_max_size:
        c.extend(e)
    
    elif rest < 0 and len(c) < c_max_size:
        left = e[:rest]
        right = e[rest:]
        
        c.extend(left)
        new_a.extend(right)

    elif len(c) >= c_max_size:
        new_a.extend(e)

print(c)
print(new_a)
```

```python
import os
import numpy as np
import pickle
import gzip
from tqdm import tqdm

old_path = "/home/bigtech/data/verbosius/amazon/preprocess/amazon_chunkdist_1250/"
n = 25

old_dir = os.listdir(old_path) 

new_path = "/home/bigtech/data/verbosius/amazon/preprocess/amazon_chunkdist_6165/"
if not os.path.exists(new_path):
    os.mkdir(new_path)

np.random.shuffle(old_dir)

old_dir = old_dir[:n]

for i, file in enumerate(tqdm(old_dir)):

    with gzip.open(os.path.join(old_path, file), "rb") as f:
        data = pickle.load(f)

    with gzip.open(os.path.join(new_path, f"init_chunk_{i}_.pkl"), "wb") as f:
        pickle.dump(data, f)
```
