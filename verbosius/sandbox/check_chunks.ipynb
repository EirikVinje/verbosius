```python
import pickle
import numpy as np
import os
from collections import Counter

path = "~/data/verbosius/amazon/chunking/amazon_chunkdist_1250/train"
path = os.path.expanduser(path)

listdir = os.listdir(path)

for i in range(20):

    chunk = np.random.choice(listdir)
    chunk = pickle.load(open(os.path.join(path, chunk), "rb"))

    print(Counter(chunk["train_y"]))
```
