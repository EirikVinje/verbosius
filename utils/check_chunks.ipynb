```python
import os
import gzip
import pickle


partdir = "/home/bigtech/data/verbosius/amazon/trainingdata/part_6165"

traindir = os.path.join(partdir, "train")
evaldir = os.path.join(partdir, "eval")

print("train:")
for tf in os.listdir(traindir):
    with gzip.open(os.path.join(traindir, tf)) as f:
        data = pickle.load(f)
        print(len(data))

print()
print("eval:")
for ef in os.listdir(evaldir):
    with gzip.open(os.path.join(evaldir, ef)) as f:
        data = pickle.load(f)
        print(len(data))
```
