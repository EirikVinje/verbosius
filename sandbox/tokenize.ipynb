```python
from tqdm import tqdm
import gzip
import numpy as np
import os
import pickle
```

```python
from xai_transformer.helper_functions import tokenize_and_align_labels
import config as cfg
```

```python
inputdir = "/home/bigtech/data/verbosius/amazon/trainingdata/amazon_chunkdist_9999"
outputdir = "/home/bigtech/data/verbosius/amazon/tokenized_trainingdata/amazon_chunkdist_9999"

chunks = os.listdir(inputdir)

for chunk_file in tqdm(chunks):

    with gzip.open(os.path.join(inputdir, chunk_file), 'rb') as f:
        chunk = pickle.load(f)
        
    tokenized_chunk = tokenize_and_align_labels(chunk, cfg.tokenizer, orig_labels=True)

    chunk_file = f"tokenized_{chunk_file}"

    with gzip.open(os.path.join(outputdir, chunk_file), 'wb') as f:
        pickle.dump(tokenized_chunk, f)
```
